## 2. The question: decision, action, cost of a wrong call

### Research Question

Which content pages should be reviewed first for possible refresh or improvement, based on observable search and engagement signals?

**Decision:** Decide which pages deserve limited review capacity first.

**Action:** A content or SEO reviewer can inspect the highest-ranked pages and decide whether to refresh, expand, protect, prune, or monitor them.

**Cost of a wrong recommendation:** A false positive could waste review and editorial time on a page that does not need attention. A false negative could cause a potentially important or declining page to be missed. Therefore, the goal is to improve the order in which pages are reviewed rather than guarantee that a page will recover.

## 3. Quick look at the data (2-3 real numbers)

```python
import os
import pandas as pd

repo_path = "/content/flyrank-ml-internship-starter"
csv_path = f"{repo_path}/data/raw/content_refresh_anonymized.csv"

# Clone the starter repository if it is not already present
if not os.path.exists(repo_path):
    !git clone https://github.com/flyrank-bih/flyrank-ml-internship-starter.git

# Load the actual FlyRank dataset
df = pd.read_csv(csv_path)

print("Number of pages:", len(df))
print("Number of columns:", len(df.columns))

declining_rate = (
    df["trend_direction"].str.lower() == "down"
).mean()

print("Declining-page rate:", round(declining_rate, 3))

visible_pages = (df["impressions_90d"] >= 500).sum()

print("Pages with 500+ impressions:", visible_pages)
```

### Observation

The starter dataset contains a substantial number of content pages, and a measurable share of pages are showing a downward trend. There are also pages receiving meaningful search exposure.

This suggests that there is value in prioritizing pages for human review instead of treating every page equally. The next step is to investigate whether observable page-level signals can support a useful review ranking.
